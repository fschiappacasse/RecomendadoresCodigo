## 2. Carga de datos y artefactos

```python
import pandas as pd
import numpy as np
import pickle
from scipy.sparse import load_npz
import kagglehub

kaggle_path = kagglehub.dataset_download("odedgolden/movielens-1m-dataset")

ratings = pd.read_csv(f"{kaggle_path}/ratings.dat", sep="::", engine="python",
                      names=["user_id", "movie_id", "rating", "timestamp"])
movies  = pd.read_csv(f"{kaggle_path}/movies.dat", sep="::", engine="python",
                      names=["movie_id", "title", "genres"], encoding="latin-1")

with open("encoders/le_user.pkl", "rb") as f:
    le_user = pickle.load(f)
with open("encoders/le_item.pkl", "rb") as f:
    le_item = pickle.load(f)

genre_features  = load_npz("encoders/genre_features.npz").toarray()   # (n_items, 18)
item_table      = pd.read_csv("item_table.csv")

text_emb  = np.load("embeddings/text_embeddings.npy")
img_emb   = np.load("embeddings/image_embeddings.npy")
fused_emb = np.load("embeddings/fused_embeddings.npy")

ratings['user_idx'] = le_user.transform(ratings['user_id'])
ratings['item_idx'] = le_item.transform(ratings['movie_id'])
n_users = len(le_user.classes_)
n_items = len(le_item.classes_)
item_ids_ordered = le_item.classes_

print(f"Usuarios: {n_users}  Items: {n_items}  Interacciones: {len(ratings)}")
```

## 3. Construcción de embeddings por ítem

```python
it = item_table.set_index('movie_id')

def build_emb_matrix(emb_array, dim):
    mat = np.zeros((n_items, dim))
    coverage = np.zeros(n_items, dtype=bool)
    for item_idx, movie_id in enumerate(item_ids_ordered):
        if movie_id in it.index:
            emb_idx = it.loc[movie_id, 'emb_idx']
            mat[item_idx] = emb_array[emb_idx]
            coverage[item_idx] = True
    return mat, coverage

text_mat,  text_cov  = build_emb_matrix(text_emb,  512)
fused_mat, fused_cov = build_emb_matrix(fused_emb, 512)
img_mat, img_cov = build_emb_matrix(img_emb, 512) #

print(f"text_mat:  {text_mat.shape}   cobertura: {text_cov.mean():.1%} de items con embedding real")
print(f"fused_mat: {fused_mat.shape}   cobertura: {fused_cov.mean():.1%} de items con embedding real")
print(f"fused_mat: {img_mat.shape}   cobertura: {fused_cov.mean():.1%} de items con embedding real")

from sklearn.preprocessing import normalize

# normalizamos solo las filas con cobertura real (las filas en cero se quedan en cero)
text_mat_norm  = text_mat.copy()
fused_mat_norm = fused_mat.copy()
img_mat_norm = img_mat.copy()

text_mat_norm[text_cov]   = normalize(text_mat[text_cov])
fused_mat_norm[fused_cov] = normalize(fused_mat[fused_cov])
img_mat_norm[img_cov] = normalize(img_mat[img_cov])


text_mat  = text_mat_norm
fused_mat = fused_mat_norm
img_mat = img_mat_norm

print(f"text_mat final:  {text_mat.shape}")
print(f"fused_mat final: {fused_mat.shape}")
```

## 4. Preparación del dataset para DeepCTR-Torch
`deepctr-torch` recibe un diccionario `{nombre_feature: array}`. Se definen las features como:
- **SparseFeat**: IDs categóricos (user, item) → aprende embeddings internos
- **DenseFeat**: features continuas (géneros, embeddings de texto/imagen)

```python
from deepctr_torch.inputs import SparseFeat, DenseFeat, get_feature_names
from sklearn.model_selection import train_test_split


ratings['label'] = (ratings['rating'] >= 4).astype(np.float32)

def build_input(ratings_df, extra_item_feat=None):
    item_idxs = ratings_df['item_idx'].values
    user_idxs = ratings_df['user_idx'].values

    sparse_cols = [
        SparseFeat('user_idx', vocabulary_size=n_users, embedding_dim=16),
        SparseFeat('item_idx', vocabulary_size=n_items, embedding_dim=16),
    ]
    dense_cols = [DenseFeat('genre_feat', dimension=genre_features.shape[1])]

    if extra_item_feat is not None:
        dense_cols.append(DenseFeat('extra_feat', dimension=extra_item_feat.shape[1]))

    feature_columns = sparse_cols + dense_cols

    # Indexar directamente 
    x = {
        'user_idx':   user_idxs.astype(np.int32),
        'item_idx':   item_idxs.astype(np.int32),
        'genre_feat': genre_features[item_idxs].astype(np.float32),  # (N, 18)
    }
    if extra_item_feat is not None:
        x['extra_feat'] = extra_item_feat[item_idxs].astype(np.float32)

    y = ratings_df['label'].values
    return x, y, feature_columns

x_base, y_base, fc_base = build_input(ratings)
print(f"base  — {len(y_base):,} muestras | RAM aprox: {x_base['genre_feat'].nbytes / 1e6:.0f} MB (generos)")

x_text, y_text, fc_text = build_input(ratings, text_mat)
print(f"text  — {len(y_text):,} muestras | RAM aprox: {x_text['extra_feat'].nbytes / 1e6:.0f} MB (text emb float32, 64d)")

x_fuse, y_fuse, fc_fuse = build_input(ratings, fused_mat)
print(f"fused — {len(y_fuse):,} muestras | RAM aprox: {x_fuse['extra_feat'].nbytes / 1e6:.0f} MB (fused emb float32, 64d)")

x_img, y_img, fc_img = build_input(ratings, img_mat)
print(f"img   — {len(y_img):,} muestras | RAM aprox: {x_img['extra_feat'].nbytes / 1e6:.0f} MB (image emb)")

print("Datasets listos:")
print(f"  base  — {len(ratings)} muestras, {len(fc_base)} feature groups")
print(f"  text  — {len(ratings)} muestras, {len(fc_text)} feature groups")
print(f"  fused — {len(ratings)} muestras, {len(fc_fuse)} feature groups")
print(f"  img — {len(ratings)} muestras, {len(img_mat)} feature groups")

```

## 5. Split Train / Test

```python
def split_x(x, y, test_size=0.2, seed=42):
    idx_tr, idx_te = train_test_split(np.arange(len(y)),
                                      test_size=test_size, random_state=seed)
    x_tr = {k: v[idx_tr] for k, v in x.items()}
    x_te = {k: v[idx_te] for k, v in x.items()}
    return x_tr, x_te, y[idx_tr], y[idx_te], idx_tr, idx_te

xb_tr, xb_te, yb_tr, yb_te, idx_tr, idx_te = split_x(x_base, y_base)
xt_tr, xt_te, yt_tr, yt_te, _,      _      = split_x(x_text, y_text)
xf_tr, xf_te, yf_tr, yf_te, _,      _      = split_x(x_fuse, y_fuse)
xi_tr, xi_te, yi_tr, yi_te, _, _ = split_x(x_img, y_img)

# Items vistos en train por usuario 
train_seen = {}
for u, i in zip(xb_tr['user_idx'], xb_tr['item_idx']):
    train_seen.setdefault(int(u), set()).add(int(i))

print(f"Train: {len(yb_tr)}  Test: {len(yb_te)}")
```

## 6. Entrenamiento con DeepCTR-Torch
Se instancia `DeepFM` directamente desde la librería. Solo hay que pasar las `feature_columns` — ella se encarga de la arquitectura completa.

```python
import torch
from deepctr_torch.models import DeepFM

DEVICE = "cuda" if torch.cuda.is_available() else "cpu"

def train_deepfm(x_tr, y_tr, feature_columns, label):
    # FIX: l2_reg_dnn configurable -- los modelos con embeddings extra (texto/imagen)
    # tienen muchos mas parametros de entrada y se benefician de mas regularizacion
    # para que el DNN no "apague" la rama nueva ni sobreajuste con ella.
    model = DeepFM(
        linear_feature_columns=feature_columns,
        dnn_feature_columns=feature_columns,
        dnn_hidden_units=(256, 128, 64),
        dnn_dropout=0.2,
        l2_reg_dnn=1e-4,
        task='binary',
        device=DEVICE,
    )
    model.compile(optimizer="adam",
                  loss="binary_crossentropy",
                  metrics=["binary_crossentropy", "auc"])

    print(f"\n--- Entrenando: {label} ---")
    model.fit(x_tr, y_tr,
              batch_size=4096,
              epochs=40,
              verbose=2,
              validation_split=0.1)
    return model

# FIX: mas regularizacion (l2_reg_dnn) y mas epocas para los modelos con
# embeddings extra -- tienen mas parametros de entrada y necesitan mas
# entrenamiento/regularizacion para que la rama nueva aporte señal real.
dfm_base  = train_deepfm(xb_tr, yb_tr, fc_base, "base (generos)")
dfm_text  = train_deepfm(xt_tr, yt_tr, fc_text, "base + sinopsis")
dfm_fused = train_deepfm(xf_tr, yf_tr, fc_fuse, "base + sinopsis + imagen")
dfm_img = train_deepfm(xi_tr, yi_tr, fc_img, "base + imagen")
```

```python
def precision_recall_at_k(df_user, k):
    df_user = df_user.sort_values('score', ascending=False)
    topk    = df_user.head(k)
    hits_k  = int(topk['label'].sum())
    total_rel = int(df_user['label'].sum())

    prec = hits_k / k
    rec  = hits_k / total_rel if total_rel > 0 else np.nan

    return prec, rec, total_rel


def macro_precision_recall_at_k(df, k):
    precs, recs = [], []
    for uid, df_u in df.groupby('user_id'):
        p, r, tot_rel = precision_recall_at_k(df_u, k)
        precs.append(p)
        if not np.isnan(r):
            recs.append(r)
    return float(np.mean(precs)), (float(np.mean(recs)) if len(recs) > 0 else np.nan)


def eval_accuracy(model, x_te, y_te, label, k=10):
    """
    Construye un DataFrame con columnas [user_id, item_id, score, label]
    y aplica macro_precision_recall_at_k.
    """
    scores = model.predict(x_te, batch_size=8192).squeeze()

    df_eval = pd.DataFrame({
        'user_id': x_te['user_idx'].astype(int),
        'item_id': x_te['item_idx'].astype(int),
        'score':   scores,
        'label':   y_te.astype(int),
    })

    p10, r10 = macro_precision_recall_at_k(df_eval, k=k)
    print(f"[DeepFM] {label:30s}  P@{k}={p10:.4f}  R@{k}={r10:.4f}")
    return p10, r10


p_base, r_base = eval_accuracy(dfm_base,  xb_te, yb_te, "base (generos)")
p_text, r_text = eval_accuracy(dfm_text,  xt_te, yt_te, "base + sinopsis")
p_full, r_full = eval_accuracy(dfm_fused, xf_te, yf_te, "base + sinopsis + imagen")
p_img, r_img = eval_accuracy(dfm_img, xi_te, yi_te, "base + imagen")
```

```python
def precision_recall_at_k(model, x_te, y_te, user_col, item_col, k=10):
    preds     = model.predict(x_te, batch_size=8192).squeeze()
    users     = user_col
    items     = item_col
    user_pred = {}
    user_true = {}
    for u, it_id, p, label in zip(users, items, preds, y_te):
        user_pred.setdefault(int(u), []).append((float(p), int(it_id)))
        if label == 1:
            user_true.setdefault(int(u), set()).add(int(it_id))
    precs, recs = [], []
    for u, scored_items in user_pred.items():
        top_ids  = {i for _, i in sorted(scored_items, reverse=True)[:k]}
        relevant = user_true.get(u, set())
        if not relevant:
            continue
        hits = len(top_ids & relevant)
        precs.append(hits / k)
        recs.append(hits / len(relevant))
    return np.mean(precs), np.mean(recs)

def eval_accuracy(model, x_te, y_te, label):
    p10, r10 = precision_recall_at_k(
        model, x_te, y_te,
        user_col=x_te['user_idx'],
        item_col=x_te['item_idx'],
    )
    print(f"[DeepFM] {label:30s}  P@10={p10:.4f}  R@10={r10:.4f}")
    return p10, r10

p_base, r_base = eval_accuracy(dfm_base,  xb_te, yb_te, "base (generos)")
p_text, r_text = eval_accuracy(dfm_text,  xt_te, yt_te, "base + sinopsis")
p_full, r_full = eval_accuracy(dfm_fused, xf_te, yf_te, "base + sinopsis + imagen")
p_img, r_img = eval_accuracy(dfm_img, xi_te, yi_te, "base + imagen")
```

## 8. Métricas de Diversidad y Novedad


Se implementan las tres métricas según las definiciones del curso:
Cita IA: https://claude.ai/share/f19c8cb6-6882-44af-8b06-cdf672f7f409


```python
import numpy as np
from sklearn.metrics.pairwise import cosine_similarity

# Popularidad de cada item
item_counts     = np.bincount(ratings['item_idx'].values, minlength=n_items).astype(float)
item_popularity = item_counts / item_counts.sum()

def ils_ziegler(top_k_lists, item_emb_matrix):
    """ILS(P_w) = [ΣΣ c_o(b_k, b_c)] / 2  —  similitud coseno intra-lista."""
    scores = []
    for user, items in top_k_lists.items():
        if len(items) < 2:
            continue
        vecs  = item_emb_matrix[items]
        sim   = cosine_similarity(vecs)
        n     = len(items)
        total = (sim.sum() - np.trace(sim)) / 2   # excluye diagonal
        scores.append(total)
    return np.mean(scores)

def novedad_self_info(top_k_lists, item_popularity):
    """Novedad = (1/|U|) Σ_u (1/L) Σ_i log(1/pop_i)."""
    user_novs = []
    for user, items in top_k_lists.items():
        pops = np.clip(item_popularity[items], 1e-10, None)
        user_novs.append(np.mean(np.log(1.0 / pops)))
    return np.mean(user_novs)

def diversidad_lathia(top_k_lists, train_seen, k=10):
    """diversity(L1, L2, N) = |L2 \ L1| / N."""
    scores = []
    for user, l2_items in top_k_lists.items():
        l1     = train_seen.get(user, set())
        nuevos = set(l2_items) - l1
        scores.append(len(nuevos) / k)
    return np.mean(scores)

print("Funciones de metricas definidas.")
```

### 8.1 Generación de listas top-K
Para cada usuario se puntúan todos los ítems y se excluyen los vistos en train.

```python
from deepctr_torch.inputs import get_feature_names

def get_top_k_deepfm(model, feature_columns, extra_mat=None, k=10, n_sample_users=2000, seed=42):
    
    rng      = np.random.default_rng(seed)
    all_u    = list(train_seen.keys())
    sampled  = rng.choice(all_u, size=min(n_sample_users, len(all_u)), replace=False)
    feat_names = get_feature_names(feature_columns)

    top_k = {}
    all_items  = np.arange(n_items)

    for u in sampled:
        user_col  = np.full(n_items, u, dtype=np.int32)
        item_col  = all_items.astype(np.int32)
        genre_col = genre_features.astype(np.float32)

        x_pred = {'user_idx': user_col,
                  'item_idx': item_col,
                  'genre_feat': genre_col}

        # Agregar extra_feat 
        if 'extra_feat' in feat_names:
            assert extra_mat is not None, "este modelo necesita extra_mat"
            x_pred['extra_feat'] = extra_mat.astype(np.float32)

        scores = model.predict(x_pred, batch_size=2048).squeeze()

        seen = list(train_seen.get(u, set()))
        if seen:
            scores[seen] = -np.inf
        top_k[u] = np.argsort(scores)[::-1][:k].tolist()

    return top_k

K = 10
print("Generando listas top-K...")
topk_base  = get_top_k_deepfm(dfm_base,  fc_base, extra_mat=None,      k=K)
topk_text  = get_top_k_deepfm(dfm_text,  fc_text, extra_mat=text_mat,  k=K)
topk_fused = get_top_k_deepfm(dfm_fused, fc_fuse, extra_mat=fused_mat, k=K)
topk_img = get_top_k_deepfm(dfm_img, fc_img, extra_mat=img_mat, k=K)
print("Listo.")
```

### 8.2 Cálculo de ILS, Novedad y Diversidad Lathia

```python
results = {}
for label, topk in [("base (generos)",           topk_base),
                    ("base + sinopsis",           topk_text),
                    ("base + imagen",             topk_img),
                    ("base + sinopsis + imagen",  topk_fused)]:
    ils  = ils_ziegler(topk, fused_mat)
    nov  = novedad_self_info(topk, item_popularity)
    lat  = diversidad_lathia(topk, train_seen, k=K)
    results[label] = {"ILS": ils, "Novedad": nov, "Lathia": lat}
    print(f"[DeepFM] {label:30s}  ILS={ils:.4f}  Novedad={nov:.4f}  Lathia={lat:.4f}")
    
```

```python
# --- Recomendación Top-K para un usuario dado (por user_id real o user_idx interno) ---

def titulo(item_idx):
    movie_id = item_ids_ordered[item_idx]
    row = movies.loc[movies['movie_id'] == movie_id, 'title']
    return row.values[0] if len(row) else f"movie_id={movie_id}"

# modelos disponibles y su config (modelo, feature_columns, extra_mat)
MODELOS = {
    "base":   (dfm_base,  fc_base, None),
    "text":   (dfm_text,  fc_text, text_mat),
    "img":    (dfm_img,   fc_img,  img_mat),
    "fused":  (dfm_fused, fc_fuse, fused_mat),
}

def recomendar_con_candidatos_test(user_idx, modelo="fused", k=10, mostrar=True):
    """
    Rankea SOLO los candidatos que xb_te tiene para este usuario
    (los mismos que se usaron para calcular P@10/R@10 en tu tabla),
    no el catálogo completo. Así los hits van a aparecer.
    """
    u_arr = np.array(xb_te['user_idx'])
    mask = (u_arr == user_idx)

    if mask.sum() == 0:
        print(f"user_idx={user_idx} no tiene filas en xb_te")
        return []

    item_candidatos = np.array(xb_te['item_idx'])[mask]
    labels_candidatos = np.array(yb_te)[mask]

    model, feature_columns, extra_mat = MODELOS[modelo]
    from deepctr_torch.inputs import get_feature_names
    feat_names = get_feature_names(feature_columns)

    n_cand = len(item_candidatos)
    x_pred = {
        'user_idx':   np.full(n_cand, user_idx, dtype=np.int32),
        'item_idx':   item_candidatos.astype(np.int32),
        'genre_feat': genre_features[item_candidatos].astype(np.float32),
    }
    if 'extra_feat' in feat_names:
        x_pred['extra_feat'] = extra_mat[item_candidatos].astype(np.float32)

    scores = model.predict(x_pred, batch_size=2048).squeeze()
    if scores.ndim == 0:
        scores = np.array([scores])

    orden = np.argsort(scores)[::-1][:k]

    recomendaciones = []
    if mostrar:
        user_id_real = le_user.inverse_transform([user_idx])[0]
        print(f"Usuario (idx={user_idx}, id real={user_id_real}) — modelo: {modelo}")
        print(f"Candidatos evaluados: {n_cand} (positivos: {int(labels_candidatos.sum())})\n")

    for i in orden:
        iidx = int(item_candidatos[i])
        es_hit = bool(labels_candidatos[i] == 1)
        recomendaciones.append((titulo(iidx), float(scores[i]), es_hit))
        if mostrar:
            marca = "✅ HIT" if es_hit else "—"
            print(f"  {marca}  {titulo(iidx)}  (score={scores[i]:.4f})")

    n_hits = sum(1 for _, _, h in recomendaciones if h)
    if mostrar:
        print(f"\nTotal hits: {n_hits}/{min(k, int(labels_candidatos.sum()))}")

    return recomendaciones

```
